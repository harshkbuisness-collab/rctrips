# Voice Studio on a free Google GPU (much faster)
1. Menu **Runtime > Change runtime type > T4 GPU > Save**.
2. Press the play button on **Step 1** and choose your **VoiceStudio.zip** file. Wait until it says *Ready* (about 3-5 minutes).
3. Press play on **Step 2**. Click the link ending in **gradio.live**: the same Voice Studio page opens.
4. Keep this tab open while it works. Download your voices from the page when they are done.

Free Colab gives a few hours of GPU at a time; for very big batches, split them over several runs (finished sentences are remembered only while this session lives, so download as you go).
